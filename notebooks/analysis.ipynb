```python
import pandas as pd

df["TOTAL_SEATS"] = df["FIRST_CLASS_SEATS"] + df["BUSINESS_CLASS_SEATS"] + df["ECONOMY_SEATS"]

grouped = df.groupby(["HAUL", "TIME_OF_DAY"]).agg(
    flights=("FLIGHT_NO", "size"),
    tier1_pax=("TIER1_ELIGIBLE_PAX", "sum"),
    tier2_pax=("TIER2_ELIGIBLE_PAX", "sum"),
    tier3_pax=("TIER3_ELIGIBLE_PAX", "sum"),
    total_seats=("TOTAL_SEATS", "sum"),
)

lookup = pd.DataFrame({
    "tier1_pct_potential": grouped["tier1_pax"] / grouped["total_seats"] * 100,
    "tier2_pct": grouped["tier2_pax"] / grouped["total_seats"] * 100,
    "tier3_pct": grouped["tier3_pax"] / grouped["total_seats"] * 100,
}).round(2).reset_index()

lookup.to_csv("../data/lounge_eligibility_lookup.csv", index=False)
lookup
```

```python
def estimate_eligible(haul, time_of_day, seats, tier):
    table = pd.read_csv("../data/lounge_eligibility_lookup.csv")
    col = {1: "tier1_pct_potential", 2: "tier2_pct", 3: "tier3_pct"}[tier]
    row = table[(table["HAUL"] == haul.upper()) & (table["TIME_OF_DAY"] == time_of_day.title())]
    if row.empty:
        raise ValueError("No matching group in the lookup table")
    return round(seats * row[col].iloc[0] / 100, 1)
```

## Assumptions
- Total seats is used as a proxy for total passengers (flights assumed full).
- Future summers are assumed to look like this schedule.
- Tier 1 is potential demand only: there is no Concorde Room at Terminal 3.
- Many flights with First Class seats show 0 Tier 1 eligible passengers, with no pattern by aircraft or region. I used the eligible-passenger columns as given.
- Percentages barely differ by time of day. I kept it because flight volume, and so lounge crowding, peaks in the morning.

```python
print(estimate_eligible("short", "lunchtime", 180, 3))   
print(estimate_eligible("long", "morning", 300, 2))     
```
