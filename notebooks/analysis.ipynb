```python
from pathlib import Path

import pandas as pd


DATA_FILENAME = "British Airways Summer Schedule Dataset - Forage Data Science Task 1.xlsx"

DATA_FILE = Path.cwd() / "data" / DATA_FILENAME
if not DATA_FILE.exists():
    DATA_FILE = Path.cwd().parent / "data" / DATA_FILENAME


def load_schedule() -> pd.DataFrame:
    """Load the British Airways summer schedule from the data folder."""
    return pd.read_excel(DATA_FILE, sheet_name=0)

df = load_schedule()
df.head()
```
